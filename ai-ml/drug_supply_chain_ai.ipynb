```python
import random
from datetime import datetime, timedelta

random.seed(42)
np.random.seed(42)

medicines = {
    "Paracetamol": 500,
    "Amoxicillin": 250,
    "Insulin": 100,
    "Azithromycin": 200,
    "Ibuprofen": 300
}

locations = [
    "Delhi",
    "Chandigarh",
    "Amritsar",
    "Ludhiana",
    "Patiala"
]

suppliers = [
    "Supplier_A",
    "Supplier_B",
    "Supplier_C"
]

data = []

start_date = datetime(2025, 1, 1)

for i in range(365):

    date = start_date + timedelta(days=i)

    for medicine, base_demand in medicines.items():

        for location in locations:

            # Weekly pattern
            day_of_week = date.weekday()

            if day_of_week >= 5:
                weekly_factor = 0.85
            else:
                weekly_factor = 1.0

            # Seasonal pattern
            month = date.month

            if month in [11, 12, 1, 2]:
                seasonal_factor = 1.15
            elif month in [6, 7, 8]:
                seasonal_factor = 0.95
            else:
                seasonal_factor = 1.0

            # Random variation
            noise = np.random.normal(0, 30)

            demand = (
                base_demand
                * weekly_factor
                * seasonal_factor
                + noise
            )

            demand = max(20, int(demand))

            # Inventory
            inventory = random.randint(
                demand * 3,
                demand * 10
            )

            # Supplier
            supplier = random.choice(suppliers)

            # Normal delivery time
            delivery_time = random.randint(2, 7)

            data.append([
                date,
                medicine,
                location,
                demand,
                inventory,
                supplier,
                delivery_time
            ])

real_df = pd.DataFrame(data, columns=[
    "Date",
    "Medicine",
    "Location",
    "Demand",
    "Inventory",
    "Supplier",
    "Delivery_Time"
])

real_df.head()
```

```python
real_df["Year"] = real_df["Date"].dt.year
real_df["Month"] = real_df["Date"].dt.month
real_df["Day"] = real_df["Date"].dt.day
real_df["DayOfWeek"] = real_df["Date"].dt.dayofweek
real_df["IsWeekend"] = (real_df["DayOfWeek"] >= 5).astype(int)

real_df.head()
```

```python
real_df = real_df.sort_values(
    ["Medicine", "Location", "Date"]
).reset_index(drop=True)
```

```python
real_df["Previous_Demand"] = (
    real_df
    .groupby(["Medicine", "Location"])["Demand"]
    .shift(1)
)
```

```python
real_df["7_Day_Average"] = (
    real_df
    .groupby(["Medicine", "Location"])["Demand"]
    .transform(
        lambda x: x.shift(1).rolling(7).mean()
    )
)
```

```python
real_df[
    [
        "Date",
        "Medicine",
        "Location",
        "Demand",
        "Previous_Demand",
        "7_Day_Average"
    ]
].head(15)
```

```python
real_df = real_df.dropna().reset_index(drop=True)
```

```python
real_df.shape
```

```python
features = [
    "Inventory",
    "Delivery_Time",
    "Year",
    "Month",
    "Day",
    "DayOfWeek",
    "IsWeekend",
    "Previous_Demand",
    "7_Day_Average"
]

target = "Demand"
```

```python
X = real_df[features]
y = real_df[target]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
split_index = int(len(real_df) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
model.fit(X_train, y_train)
```

```python
predictions = model.predict(X_test)

predictions[:10]
```

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": predictions
})

comparison.head(10)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y_test, predictions)

print("Mean Absolute Error:", mae)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    y_test.values[:100],
    label="Actual Demand"
)

plt.plot(
    predictions[:100],
    label="Predicted Demand"
)

plt.xlabel("Test Sample")
plt.ylabel("Demand")
plt.title("Actual vs Predicted Medicine Demand")

plt.legend()
plt.show()
```
